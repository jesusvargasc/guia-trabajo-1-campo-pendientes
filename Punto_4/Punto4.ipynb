# Punto 4 — Modelo de población

La tasa de cambio de la población de cierta especie está dada por:

$$
\frac{dP}{dt}=3P-2P^2
$$

donde $P$ representa la población en miles de ejemplares y $t$
representa el tiempo en años.

Para analizar el comportamiento de la población utilizaremos los
puntos críticos y el diagrama de fase.

## 4(a) — Diagrama de fase

La ecuación diferencial es:

$$
P'=3P-2P^2
$$

Factorizamos:

$$
P'=P(3-2P)
$$

Los puntos críticos se obtienen haciendo:

$$
P(3-2P)=0
$$

Por lo tanto:

$$
P=0
$$

o

$$
3-2P=0
$$

de donde:

$$
P=\frac{3}{2}
$$

Así, los puntos críticos son:

$$
P=0,\qquad P=\frac{3}{2}
$$

Estos puntos dividen la recta real en los intervalos:

$$
(-\infty,0),\qquad
\left(0,\frac{3}{2}\right),\qquad
\left(\frac{3}{2},\infty\right)
$$

```python
# ============================================================
# PUNTO 4(a)
# P' = 3P - 2P^2
# ============================================================

import numpy as np
import matplotlib.pyplot as plt

def g_4(P):
    return 3*P - 2*P**2


# Puntos críticos
puntos_criticos_4 = [0, 1.5]

print("Puntos críticos:")

for punto in puntos_criticos_4:
    print(f"P = {punto}")


# Intervalos
intervalos_4 = [
    (-1, 0),
    (0, 1.5),
    (1.5, 3)
]

print("\nAnálisis de signos:")

for a, b in intervalos_4:

    punto_prueba = (a + b) / 2
    valor = g_4(punto_prueba)

    if valor > 0:
        direccion = "↑ Creciente"
    else:
        direccion = "↓ Decreciente"

    print(
        f"Intervalo ({a}, {b}): "
        f"g(P) = {valor:.2f} → {direccion}"
    )
```

```python
# ============================================================
# DIAGRAMA DE FASE - PUNTO 4(a)
# ============================================================

fig, ax = plt.subplots(figsize=(8, 6))

# Línea de fase
ax.axvline(0, linewidth=2)

# ------------------------------------------------------------
# P < 0 → decrece
# ------------------------------------------------------------

ax.annotate(
    '',
    xy=(0, -1),
    xytext=(0, -3),
    arrowprops=dict(
        arrowstyle='->',
        linewidth=3
    )
)

# ------------------------------------------------------------
# 0 < P < 1.5 → crece
# ------------------------------------------------------------

ax.annotate(
    '',
    xy=(0, 1.2),
    xytext=(0, 0.3),
    arrowprops=dict(
        arrowstyle='->',
        linewidth=3
    )
)

# ------------------------------------------------------------
# P > 1.5 → decrece
# ------------------------------------------------------------

ax.annotate(
    '',
    xy=(0, 1.8),
    xytext=(0, 3),
    arrowprops=dict(
        arrowstyle='->',
        linewidth=3
    )
)

# Puntos críticos
for punto in puntos_criticos_4:
    ax.scatter(
        0,
        punto,
        s=150,
        zorder=5
    )

# Etiquetas
ax.text(
    0.08,
    0,
    'P = 0'
)

ax.text(
    0.08,
    1.5,
    'P = 1.5'
)

# Configuración
ax.set_ylim(-3, 3)
ax.set_xlim(-1, 1)

ax.set_xticks([])
ax.set_ylabel('P (miles)')

ax.set_title("Diagrama de fase - P' = 3P - 2P²")

ax.grid(axis='y', alpha=0.2)

plt.show()
```

## Análisis del diagrama de fase

Para $P<0$, las soluciones son decrecientes.

Para $0<P<1.5$, las soluciones son crecientes.

Para $P>1.5$, las soluciones son decrecientes.

En $P=0$, las soluciones se alejan del punto por la derecha, por lo
que:

$$
P=0
$$

es un punto inestable.

En $P=1.5$, las flechas de ambos lados apuntan hacia el punto crítico,
por lo que:

$$
P=1.5
$$

es un punto asintóticamente estable.

## 4(b) — Población inicial de 2000 ejemplares

La población inicial es:

$$
P(0)=2000
$$

Como $P$ está expresada en miles:

$$
P(0)=2
$$

Como:

$$
2>1.5
$$

la población se encuentra en el intervalo $P>1.5$.

En este intervalo:

$$
P'<0
$$

por lo que la población disminuye.

De acuerdo con el diagrama de fase, la población se aproxima al
punto de equilibrio:

$$
P=1.5
$$

Por lo tanto:

$$
\lim_{t\to\infty}P(t)=1.5
$$

La población después de mucho tiempo se aproxima a:

$$
1500\text{ ejemplares}
$$

## 4(c) — Población inicial de 100 ejemplares

La población inicial es:

$$
P(0)=100
$$

Como $P$ está expresada en miles:

$$
P(0)=0.1
$$

Se tiene:

$$
0<0.1<1.5
$$

por lo que la población se encuentra en el intervalo $(0,1.5)$.

En este intervalo:

$$
P'>0
$$

por lo que la población crece.

La población se aproxima al punto de equilibrio:

$$
P=1.5
$$

Por lo tanto:

$$
\lim_{t\to\infty}P(t)=1.5
$$

La población después de mucho tiempo se aproxima a:

$$
1500\text{ ejemplares}
$$

## 4(d) — Población inicial de 1500 ejemplares

La población inicial es:

$$
P(0)=1500
$$

Como $P$ está expresada en miles:

$$
P(0)=1.5
$$

Este valor corresponde a un punto crítico de la ecuación:

$$
P'=P(3-2P)
$$

Al evaluar:

$$
P'=1.5(3-2(1.5))=0
$$

Por lo tanto, la población no cambia con el tiempo.

Así:

$$
P(t)=1.5
$$

para todo $t$.

La población permanece constante en:

$$
1500\text{ ejemplares}
$$

## 4(e) — Tasa de nacimientos y muertes

La guía indica que nacen 150 ejemplares por trimestre y mueren $s$
ejemplares durante el mismo periodo.

La tasa neta de cambio por trimestre es:

$$
150-s
$$

Como un año tiene cuatro trimestres, la tasa anual es:

$$
4(150-s)=600-4s
$$

Si la población se expresa en ejemplares, entonces:

$$
\frac{dP}{dt}=600-4s
$$

Como en los ejercicios anteriores la población se expresa en miles,
la tasa equivalente en miles de ejemplares por año es:

$$
\frac{dP}{dt}=\frac{600-4s}{1000}
$$

## 4(f) — Análisis del comportamiento

La tasa anual de cambio de la población es:

$$
\frac{dP}{dt}=\frac{600-4s}{1000}
$$

El comportamiento depende del valor de $s$:

### Si $s<150$

Entonces:

$$
600-4s>0
$$

por lo que:

$$
\frac{dP}{dt}>0
$$

La población aumenta.

### Si $s=150$

Entonces:

$$
600-4s=0
$$

por lo que:

$$
\frac{dP}{dt}=0
$$

La población permanece constante.

### Si $s>150$

Entonces:

$$
600-4s<0
$$

por lo que:

$$
\frac{dP}{dt}<0
$$

La población disminuye.

Por lo tanto, el comportamiento de la población depende de la
relación entre la cantidad de nacimientos y la cantidad de muertes
por trimestre.
