# Punto 3 — Modelo de población

Sea $P(t)$ la población de cierta especie en un parque natural,
donde $t$ representa el tiempo en años y $P$ representa la población
en miles de ejemplares.

La ecuación diferencial que describe la tasa de cambio de la
población es:

$$
\frac{dP}{dt}=P(P-1)(2-P)
$$

Para analizar el comportamiento de la población utilizaremos el
diagrama de fase y los puntos críticos de la ecuación.

## 3(a) Diagrama de fase

La ecuación es:

$$
P'=P(P-1)(2-P)
$$

Los puntos críticos se obtienen haciendo:

$$
P(P-1)(2-P)=0
$$

Por lo tanto:

$$
{P=0,\quad P=1,\quad P=2}
$$

Estos puntos dividen la recta de fase en los intervalos:

$$
(-\infty,0),\quad (0,1),\quad (1,2),\quad (2,\infty)
$$

```python
# ============================================================
# PUNTO 3(a)
# Modelo de población
# P' = P(P-1)(2-P)
# ============================================================

def g_p(P):
    return P * (P - 1) * (2 - P)


# Puntos críticos
puntos_criticos_p = [0, 1, 2]

print("Puntos críticos:")

for punto in puntos_criticos_p:
    print(f"P = {punto}")

# Intervalos
intervalos_p = [
    (-1, 0),
    (0, 1),
    (1, 2),
    (2, 3)
]

print("\nAnálisis de signos:")

for a, b in intervalos_p:

    punto_prueba = (a + b) / 2
    valor = g_p(punto_prueba)

    if valor > 0:
        direccion = "↑ Creciente"
    else:
        direccion = "↓ Decreciente"

    print(
        f"Intervalo ({a}, {b}): "
        f"g(P) = {valor:.2f} → {direccion}"
    )
```

```python
# ============================================================
# DIAGRAMA DE FASE - PUNTO 3(a)
# ============================================================

import numpy as np
import matplotlib.pyplot as plt

# Puntos críticos
puntos_criticos_p = [0, 1, 2]

# Crear figura
fig, ax = plt.subplots(figsize=(8, 6))

# Línea de fase
ax.axvline(0, linewidth=2)

# ------------------------------------------------------------
# P < 0 → decrece
# ------------------------------------------------------------

ax.annotate(
    '',
    xy=(0, -1),
    xytext=(0, -3),
    arrowprops=dict(
        arrowstyle='->',
        linewidth=3
    )
)

# ------------------------------------------------------------
# 0 < P < 1 → crece
# ------------------------------------------------------------

ax.annotate(
    '',
    xy=(0, 0.8),
    xytext=(0, 0.2),
    arrowprops=dict(
        arrowstyle='->',
        linewidth=3
    )
)

# ------------------------------------------------------------
# 1 < P < 2 → decrece
# ------------------------------------------------------------

ax.annotate(
    '',
    xy=(0, 1.2),
    xytext=(0, 1.8),
    arrowprops=dict(
        arrowstyle='->',
        linewidth=3
    )
)

# ------------------------------------------------------------
# P > 2 → decrece
# ------------------------------------------------------------

ax.annotate(
    '',
    xy=(0, 2.5),
    xytext=(0, 4),
    arrowprops=dict(
        arrowstyle='->',
        linewidth=3
    )
)

# ------------------------------------------------------------
# Puntos críticos
# ------------------------------------------------------------

for punto in puntos_criticos_p:
    ax.scatter(
        0,
        punto,
        s=150,
        zorder=5
    )

# ------------------------------------------------------------
# Etiquetas
# ------------------------------------------------------------

ax.text(0.08, 0, 'P = 0')
ax.text(0.08, 1, 'P = 1')
ax.text(0.08, 2, 'P = 2')

# ------------------------------------------------------------
# Configuración
# ------------------------------------------------------------

ax.set_ylim(-3, 4)
ax.set_xlim(-1, 1)

ax.set_xticks([])
ax.set_ylabel('P (miles)')

ax.set_title("Diagrama de fase - P' = P(P-1)(2-P)")

ax.grid(axis='y', alpha=0.2)

plt.show()
```

## Análisis del diagrama de fase

A partir del análisis de signos:

- Para $P<0$, la población decrece.
- Para $0<P<1$, la población crece.
- Para $1<P<2$, la población decrece.
- Para $P>2$, la población decrece.

Para poblaciones físicamente posibles ($P\geq0$):

### $P=0$

Las flechas a ambos lados apuntan hacia valores menores y mayores
respectivamente, por lo que el punto es inestable.

$$
{P=0\text{ es inestable}}
$$

### $P=1$

Las flechas de ambos lados apuntan hacia $P=1$.

$$
{P=1\text{ es asintóticamente estable}}
$$

### $P=2$

Las flechas de ambos lados apuntan hacia valores menores que $2$.

Por lo tanto:

$$
{P=2\text{ es inestable}}
$$

## 3(b) — Población inicial de 3000 ejemplares

La población inicial es:

$$
P(0)=3000
$$

Como $P$ está expresada en miles de ejemplares:

$$
P(0)=3
$$

Como $3>2$, la población se encuentra en el intervalo:

$$
P>2
$$

En este intervalo:

$$
P'<0
$$

por lo que la población disminuye.

De acuerdo con el diagrama de fase, las soluciones que comienzan
por encima de $P=2$ se aproximan a $P=2$ cuando el tiempo aumenta.

Por lo tanto:

$$
{\lim_{t\to\infty}P(t)=2}
$$

Es decir, después de mucho tiempo la población se aproxima a:

$$
{2000\text{ ejemplares}}
$$

## 3(c) — Población inicial de 1500 ejemplares

La población inicial es:

$$
P(0)=1500
$$

Como $P$ está expresada en miles:

$$
P(0)=1.5
$$

Se tiene:

$$
1<1.5<2
$$

Por lo tanto, la población se encuentra en el intervalo $(1,2)$.

En este intervalo:

$$
P'<0
$$

por lo que la población disminuye.

De acuerdo con el diagrama de fase, la población se aproxima al
punto de equilibrio $P=1$.

Por lo tanto:

$$
{\lim_{t\to\infty}P(t)=1}
$$

Es decir, después de mucho tiempo la población se aproxima a:

$$
{1000\text{ ejemplares}}
$$

## 3(d) — Población inicial de 500 ejemplares

La población inicial es:

$$
P(0)=500
$$

Como $P$ está expresada en miles:

$$
P(0)=0.5
$$

Se tiene:

$$
0<0.5<1
$$

Por lo tanto, la población se encuentra en el intervalo $(0,1)$.

En este intervalo:

$$
P'>0
$$

por lo que la población crece.

De acuerdo con el diagrama de fase, la población se aproxima al
punto de equilibrio $P=1$.

Por lo tanto:

$$
{\lim_{t\to\infty}P(t)=1}
$$

Es decir, después de mucho tiempo la población se aproxima a:

$$
{1000\text{ ejemplares}}
$$

## 3(e) — ¿Puede una población de 900 ejemplares crecer hasta 1100?

La población inicial es:

$$
P(0)=900
$$

Como $P$ está expresada en miles:

$$
P(0)=0.9
$$

La población deseada sería:

$$
1100=1.1
$$

Como:

$$
0<0.9<1
$$

la población inicialmente crece porque:

$$
P'>0
$$

Sin embargo, $P=1$ es un punto de equilibrio estable. Por lo tanto,
una población que comienza por debajo de $P=1$ se aproxima a este
valor, pero no lo sobrepasa.

Así:

$$
0.9\longrightarrow1
$$

pero no:

$$
0.9\longrightarrow1.1
$$

Por lo tanto:

$$
{\text{No, una población inicial de 900 ejemplares no puede
crecer hasta 1100 ejemplares}}
$$

La población se aproxima a 1000 ejemplares.
