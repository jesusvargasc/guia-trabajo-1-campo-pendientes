## Punto 2 — Ecuaciones autónomas

### Objetivo

Analizar los puntos críticos de diferentes ecuaciones diferenciales autónomas mediante el diagrama de fase.

Para una ecuación de la forma

$$
y'=g(y)
$$

los puntos críticos se obtienen resolviendo

$$
g(y)=0.
$$

Posteriormente se analiza el signo de $g(y)$ en los intervalos determinados por los puntos críticos.

Si $g(y)>0$, las soluciones crecen.

Si $g(y)<0$, las soluciones decrecen.

Finalmente, mediante el diagrama de fase se determina la estabilidad de cada punto crítico.


# 2(a)

La ecuación diferencial autónoma es:

$$
y'=y(3-y)(y-2)
$$

## Puntos críticos

Para encontrar los puntos críticos hacemos:

$$
y(3-y)(y-2)=0
$$

Por lo tanto:

$$
\boxed{y=0,\quad y=2,\quad y=3}
$$

Estos valores dividen la recta real en los intervalos:

$$
(-\infty,0),\quad (0,2),\quad (2,3),\quad (3,\infty)
$$

```python
import numpy as np
import matplotlib.pyplot as plt

# Función de la ecuación
def g(y):
    return y * (3 - y) * (y - 2)

# Puntos críticos
puntos_criticos = [0, 2, 3]

# Intervalos
intervalos = [
    (-5, 0),
    (0, 2),
    (2, 3),
    (3, 5)
]

print("Puntos críticos:")
for punto in puntos_criticos:
    print(f"y = {punto}")

print("\nAnálisis de signos:")

for a, b in intervalos:
    punto_prueba = (a + b) / 2
    valor = g(punto_prueba)

    if valor > 0:
        direccion = "↑ Creciente"
    else:
        direccion = "↓ Decreciente"

    print(
        f"Intervalo ({a}, {b}): "
        f"g(y) = {valor:.2f} → {direccion}"
    )
```

```python
# ============================================================
# DIAGRAMA DE FASE - PUNTO 2(a)
# y' = y(3-y)(y-2)
# ============================================================

fig, ax = plt.subplots(figsize=(8, 6))

# Línea vertical que representa la recta de fase
ax.axvline(0, linewidth=2)

# ------------------------------------------------------------
# Flechas de dirección
# ------------------------------------------------------------

# y < 0  →  decrece
ax.annotate(
    '',
    xy=(0, -4),
    xytext=(0, -1),
    arrowprops=dict(
        arrowstyle='->',
        linewidth=3
    )
)

# 0 < y < 2  →  crece
ax.annotate(
    '',
    xy=(0, 1.5),
    xytext=(0, 0.5),
    arrowprops=dict(
        arrowstyle='->',
        linewidth=3
    )
)

# 2 < y < 3  →  decrece
ax.annotate(
    '',
    xy=(0, 2.3),
    xytext=(0, 2.7),
    arrowprops=dict(
        arrowstyle='->',
        linewidth=3
    )
)

# y > 3  →  crece
ax.annotate(
    '',
    xy=(0, 5),
    xytext=(0, 3.5),
    arrowprops=dict(
        arrowstyle='->',
        linewidth=3
    )
)

# ------------------------------------------------------------
# Puntos críticos
# ------------------------------------------------------------

for punto in puntos_criticos:
    ax.scatter(0, punto, s=150, zorder=5)

# ------------------------------------------------------------
# Etiquetas
# ------------------------------------------------------------

ax.text(0.08, 0, 'y = 0  →  Inestable',
        verticalalignment='center')

ax.text(0.08, 2, 'y = 2  →  Estable',
        verticalalignment='center')

ax.text(0.08, 3, 'y = 3  →  Inestable',
        verticalalignment='center')

# ------------------------------------------------------------
# Configuración
# ------------------------------------------------------------

ax.set_ylim(-5, 5)
ax.set_xlim(-1, 1)

ax.set_xticks([])
ax.set_ylabel('y')

ax.set_title("Diagrama de fase - y' = y(3-y)(y-2)")

ax.grid(axis='y', alpha=0.2)

plt.show()
```

## Análisis de estabilidad

A partir del diagrama de fase se observa el comportamiento de las soluciones
en cada intervalo.

- Si $y'<0$, las soluciones decrecen.
- Si $y'>0$, las soluciones crecen.

Para $y=0$, las flechas a ambos lados se alejan del punto crítico.
Por lo tanto:

$$
\boxed{y=0\text{ es inestable}}
$$

Para $y=2$, las flechas de ambos lados apuntan hacia el punto crítico.
Por lo tanto:

$$
\boxed{y=2\text{ es asintóticamente estable}}
$$

Para $y=3$, las flechas a ambos lados se alejan del punto crítico.
Por lo tanto:

$$
\boxed{y=3\text{ es inestable}}
$$

### Conclusión

Los puntos críticos de la ecuación son $y=0$, $y=2$ y $y=3$.
De ellos, $y=2$ es asintóticamente estable, mientras que $y=0$ y
$y=3$ son inestables.

# 2(b)

La ecuación diferencial autónoma es:

$$
y'=y^2-y^3
$$

Factorizando:

$$
y'=y^2(1-y)
$$

## Puntos críticos

Los puntos críticos se obtienen haciendo:

$$
y^2(1-y)=0
$$

Por lo tanto:

$$
\boxed{y=0,\quad y=1}
$$

Estos puntos dividen la recta real en los intervalos:

$$
(-\infty,0),\quad (0,1),\quad (1,\infty)
$$

```python
# ============================================================
# PUNTO 2(b)
# y' = y^2 - y^3
# ============================================================

def g_b(y):
    return y**2 - y**3


# Puntos críticos
puntos_criticos_b = [0, 1]

# Intervalos
intervalos_b = [
    (-5, 0),
    (0, 1),
    (1, 5)
]

print("Puntos críticos:")
for punto in puntos_criticos_b:
    print(f"y = {punto}")

print("\nAnálisis de signos:")

for a, b in intervalos_b:

    punto_prueba = (a + b) / 2
    valor = g_b(punto_prueba)

    if valor > 0:
        direccion = "↑ Creciente"
    else:
        direccion = "↓ Decreciente"

    print(
        f"Intervalo ({a}, {b}): "
        f"g(y) = {valor:.2f} → {direccion}"
    )
```

```python
# ============================================================
# DIAGRAMA DE FASE - PUNTO 2(b)
# ============================================================

fig, ax = plt.subplots(figsize=(8, 6))

# Línea de fase
ax.axvline(0, linewidth=2)

# y < 0 → crece
ax.annotate(
    '',
    xy=(0, -1),
    xytext=(0, -4),
    arrowprops=dict(
        arrowstyle='->',
        linewidth=3
    )
)

# 0 < y < 1 → crece
ax.annotate(
    '',
    xy=(0, 0.8),
    xytext=(0, 0.2),
    arrowprops=dict(
        arrowstyle='->',
        linewidth=3
    )
)

# y > 1 → decrece
ax.annotate(
    '',
    xy=(0, 1.5),
    xytext=(0, 4),
    arrowprops=dict(
        arrowstyle='->',
        linewidth=3
    )
)

# Puntos críticos
for punto in puntos_criticos_b:
    ax.scatter(0, punto, s=150, zorder=5)

# Etiquetas
ax.text(
    0.08, 0,
    'y = 0  →  Semiestable',
    verticalalignment='center'
)

ax.text(
    0.08, 1,
    'y = 1  →  Estable',
    verticalalignment='center'
)

# Configuración
ax.set_ylim(-5, 5)
ax.set_xlim(-1, 1)

ax.set_xticks([])
ax.set_ylabel('y')

ax.set_title("Diagrama de fase - y' = y² - y³")

ax.grid(axis='y', alpha=0.2)

plt.show()
```

## Análisis de estabilidad

A partir del diagrama de fase se observa que:

- Para $y<0$, las soluciones crecen.
- Para $0<y<1$, las soluciones crecen.
- Para $y>1$, las soluciones decrecen.

Para $y=0$, las soluciones se acercan al punto desde un lado y se
alejan desde el otro. Por lo tanto:

$$
\boxed{y=0\text{ es semiestable}}
$$

Para $y=1$, las flechas de ambos lados apuntan hacia el punto crítico.
Por lo tanto:

$$
\boxed{y=1\text{ es asintóticamente estable}}
$$

### Conclusión

Los puntos críticos de la ecuación son $y=0$ y $y=1$.
El punto $y=0$ es semiestable, mientras que $y=1$ es
asintóticamente estable.

# 2(c)

La ecuación diferencial autónoma es:

$$
y'=(y+2)(10+3y-y^2)
$$

## Puntos críticos

Los puntos críticos se obtienen haciendo:

$$
(y+2)(10+3y-y^2)=0
$$

Primero:

$$
y+2=0
$$

por lo que:

$$
y=-2
$$

Ahora resolvemos:

$$
10+3y-y^2=0
$$

Reorganizando:

$$
y^2-3y-10=0
$$

Factorizando:

$$
(y-5)(y+2)=0
$$

Por lo tanto:

$$
y=5
\qquad \text{o} \qquad
y=-2
$$

Así, los puntos críticos son:

$$
\boxed{y=-2,\quad y=5}
$$

Podemos factorizar completamente la ecuación:

$$
y'=(y+2)^2(5-y)
$$

Los puntos críticos dividen la recta real en los intervalos:

$$
(-\infty,-2),\quad (-2,5),\quad (5,\infty)
$$

```python
# ============================================================
# PUNTO 2(c)
# y' = (y+2)(10+3y-y^2)
# ============================================================

def g_c(y):
    return (y + 2) * (10 + 3*y - y**2)


# Puntos críticos
puntos_criticos_c = [-2, 5]

# Intervalos
intervalos_c = [
    (-6, -2),
    (-2, 5),
    (5, 8)
]

print("Puntos críticos:")

for punto in puntos_criticos_c:
    print(f"y = {punto}")

print("\nAnálisis de signos:")

for a, b in intervalos_c:

    # Punto de prueba
    punto_prueba = (a + b) / 2

    # Evaluamos la función
    valor = g_c(punto_prueba)

    if valor > 0:
        direccion = "↑ Creciente"
    else:
        direccion = "↓ Decreciente"

    print(
        f"Intervalo ({a}, {b}): "
        f"g(y) = {valor:.2f} → {direccion}"
    )
```

```python
# ============================================================
# DIAGRAMA DE FASE - PUNTO 2(c)
# ============================================================

fig, ax = plt.subplots(figsize=(8, 6))

# Línea de fase
ax.axvline(0, linewidth=2)

# ------------------------------------------------------------
# y < -2  → crece
# ------------------------------------------------------------

ax.annotate(
    '',
    xy=(0, -3),
    xytext=(0, -5),
    arrowprops=dict(
        arrowstyle='->',
        linewidth=3
    )
)

# ------------------------------------------------------------
# -2 < y < 5  → crece
# ------------------------------------------------------------

ax.annotate(
    '',
    xy=(0, 3),
    xytext=(0, 0),
    arrowprops=dict(
        arrowstyle='->',
        linewidth=3
    )
)

# ------------------------------------------------------------
# y > 5  → decrece
# ------------------------------------------------------------

ax.annotate(
    '',
    xy=(0, 4),
    xytext=(0, 7),
    arrowprops=dict(
        arrowstyle='->',
        linewidth=3
    )
)

# ------------------------------------------------------------
# Puntos críticos
# ------------------------------------------------------------

for punto in puntos_criticos_c:
    ax.scatter(0, punto, s=150, zorder=5)

# ------------------------------------------------------------
# Etiquetas
# ------------------------------------------------------------

ax.text(
    0.08, -2,
    'y = -2  →  Semiestable',
    verticalalignment='center'
)

ax.text(
    0.08, 5,
    'y = 5  →  Estable',
    verticalalignment='center'
)

# Configuración
ax.set_ylim(-6, 8)
ax.set_xlim(-1, 1)

ax.set_xticks([])
ax.set_ylabel('y')

ax.set_title("Diagrama de fase - y' = (y+2)(10+3y-y²)")

ax.grid(axis='y', alpha=0.2)

plt.show()
```

## Análisis de estabilidad

A partir del análisis de signos se obtiene:

- Para $y<-2$, las soluciones son crecientes.
- Para $-2<y<5$, las soluciones son crecientes.
- Para $y>5$, las soluciones son decrecientes.

### Punto crítico $y=-2$

Las soluciones se acercan a $y=-2$ por la izquierda, pero se alejan
por la derecha. Por lo tanto:

$$
\boxed{y=-2\text{ es semiestable}}
$$

### Punto crítico $y=5$

Las soluciones se acercan a $y=5$ tanto por la izquierda como por
la derecha. Por lo tanto:

$$
\boxed{y=5\text{ es asintóticamente estable}}
$$

### Conclusión

Los puntos críticos de la ecuación son:

$$
\boxed{y=-2,\quad y=5}
$$

El punto $y=-2$ es semiestable, mientras que $y=5$ es
asintóticamente estable.

# 2(d)

La ecuación diferencial autónoma es:

$$
y'=y^5-4y^3-5y^2
$$

## Puntos críticos

Para encontrar los puntos críticos hacemos:

$$
y^5-4y^3-5y^2=0
$$

Factorizamos:

$$
y^2(y^3-4y-5)=0
$$

Por lo tanto, uno de los puntos críticos es:

$$
y=0
$$

Ahora debemos resolver:

$$
y^3-4y-5=0
$$

Probando $y=-2$:

$$
(-2)^3-4(-2)-5
=-8+8-5=-5
$$

Probando $y=-1$:

$$
(-1)^3-4(-1)-5
=-1+4-5=-2
$$

Probando $y=1$:

$$
1-4-5=-8
$$

Probando $y=2$:

$$
8-8-5=-5
$$

Probando $y=3$:

$$
27-12-5=10
$$

Por lo tanto, la ecuación cúbica tiene una raíz real entre $2$ y $3$.

Esta raíz es aproximadamente:

$$
y\approx2.456
$$

Por lo tanto, los puntos críticos son aproximadamente:

$$
\boxed{y=0,\quad y\approx2.456}
$$

Estos puntos dividen la recta real en los intervalos:

$$
(-\infty,0),\quad (0,2.456),\quad (2.456,\infty)
$$

```python
# ============================================================
# PUNTO 2(d)
# y' = y^5 - 4y^3 - 5y^2
# ============================================================

def g_d(y):
    return y**5 - 4*y**3 - 5*y**2


# Encontramos las raíces de:
# y^5 - 4y^3 - 5y^2 = 0

coeficientes = [1, 0, -4, -5, 0, 0]

raices = np.roots(coeficientes)

print("Raíces de la ecuación:")

for raiz in raices:
    if abs(raiz.imag) < 1e-10:
        print(f"y = {raiz.real:.6f}")


# Puntos críticos reales
puntos_criticos_d = [0, 2.456]

print("\nPuntos críticos reales:")
for punto in puntos_criticos_d:
    print(f"y = {punto}")


# Intervalos de análisis
intervalos_d = [
    (-4, 0),
    (0, 2.456),
    (2.456, 5)
]

print("\nAnálisis de signos:")

for a, b in intervalos_d:

    punto_prueba = (a + b) / 2

    valor = g_d(punto_prueba)

    if valor > 0:
        direccion = "↑ Creciente"
    else:
        direccion = "↓ Decreciente"

    print(
        f"Intervalo ({a}, {b}): "
        f"g(y) = {valor:.2f} → {direccion}"
    )
```

```python
# ============================================================
# DIAGRAMA DE FASE - PUNTO 2(d)
# ============================================================

fig, ax = plt.subplots(figsize=(8, 6))

# Línea de fase
ax.axvline(0, linewidth=2)


# ------------------------------------------------------------
# y < 0 → decrece
# ------------------------------------------------------------

ax.annotate(
    '',
    xy=(0, -3),
    xytext=(0, -5),
    arrowprops=dict(
        arrowstyle='->',
        linewidth=3
    )
)


# ------------------------------------------------------------
# 0 < y < 2.456 → decrece
# ------------------------------------------------------------

ax.annotate(
    '',
    xy=(0, 1.5),
    xytext=(0, 0.5),
    arrowprops=dict(
        arrowstyle='->',
        linewidth=3
    )
)


# ------------------------------------------------------------
# y > 2.456 → crece
# ------------------------------------------------------------

ax.annotate(
    '',
    xy=(0, 5),
    xytext=(0, 3),
    arrowprops=dict(
        arrowstyle='->',
        linewidth=3
    )
)


# ------------------------------------------------------------
# Puntos críticos
# ------------------------------------------------------------

for punto in puntos_criticos_d:
    ax.scatter(
        0,
        punto,
        s=150,
        zorder=5
    )


# ------------------------------------------------------------
# Etiquetas
# ------------------------------------------------------------

ax.text(
    0.08,
    0,
    'y = 0  →  Semiestable',
    verticalalignment='center'
)

ax.text(
    0.08,
    2.456,
    'y ≈ 2.456  →  Inestable',
    verticalalignment='center'
)


# Configuración
ax.set_ylim(-5, 5)
ax.set_xlim(-1, 1)

ax.set_xticks([])
ax.set_ylabel('y')

ax.set_title(
    "Diagrama de fase - y' = y⁵ - 4y³ - 5y²"
)

ax.grid(axis='y', alpha=0.2)

plt.show()
```

## Análisis de estabilidad

A partir del análisis de signos se obtiene:

- Para $y<0$, las soluciones decrecen.
- Para $0<y<2.456$, las soluciones decrecen.
- Para $y>2.456$, las soluciones crecen.

### Punto crítico $y=0$

Las soluciones se alejan de $y=0$ por la izquierda y se acercan por
la derecha. Por lo tanto:

$$
\boxed{y=0\text{ es semiestable}}
$$

### Punto crítico $y\approx2.456$

Las soluciones se alejan de este punto tanto por la izquierda como por
la derecha. Por lo tanto:

$$
\boxed{y\approx2.456\text{ es inestable}}
$$

### Conclusión

Los puntos críticos de la ecuación son:

$$
\boxed{y=0,\quad y\approx2.456}
$$

El punto $y=0$ es semiestable y el punto $y\approx2.456$ es inestable.

# 2(e)

La ecuación diferencial autónoma es:

$$
y'=(1-y)(y-2)^3
$$

## Puntos críticos

Los puntos críticos se obtienen haciendo:

$$
(1-y)(y-2)^3=0
$$

Por lo tanto:

$$
1-y=0
$$

de donde:

$$
y=1
$$

También:

$$
(y-2)^3=0
$$

de donde:

$$
y=2
$$

Por lo tanto, los puntos críticos son:

$$
\boxed{y=1,\quad y=2}
$$

Estos puntos dividen la recta real en los intervalos:

$$
(-\infty,1),\quad (1,2),\quad (2,\infty)
$$

```python
# ============================================================
# PUNTO 2(e)
# y' = (1-y)(y-2)^3
# ============================================================

def g_e(y):
    return (1 - y) * (y - 2)**3


# Puntos críticos
puntos_criticos_e = [1, 2]

# Intervalos
intervalos_e = [
    (-4, 1),
    (1, 2),
    (2, 5)
]

print("Puntos críticos:")

for punto in puntos_criticos_e:
    print(f"y = {punto}")

print("\nAnálisis de signos:")

for a, b in intervalos_e:

    # Punto de prueba
    punto_prueba = (a + b) / 2

    # Evaluamos g(y)
    valor = g_e(punto_prueba)

    if valor > 0:
        direccion = "↑ Creciente"
    else:
        direccion = "↓ Decreciente"

    print(
        f"Intervalo ({a}, {b}): "
        f"g(y) = {valor:.2f} → {direccion}"
    )
```
